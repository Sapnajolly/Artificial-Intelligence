```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

for pkg in ['punkt', 'punkt_tab', 'stopwords', 'wordnet', 'omw-1.4']:
    nltk.download(pkg)

stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
```

```python
samples = {
    "Social Media Post": "OMG!!! Just got the new iPhone 15 Pro and it's AMAZING 😍🔥 Battery life is INSANE, no more charging every 2 hrs like my old phone lol. Definitely worth the $999... #tech #apple #blessed 🙌",
    "News Article": "The Federal Reserve announced on Wednesday that it will NOT be raising interest rates this quarter, citing 'persistent uncertainty' in the labor market. Analysts had widely expected a 0.25% hike, but the decision surprised many on Wall St. Markets reacted positively, with the S&P 500 climbing 1.2% by midday.",
    "Customer Review": "i ordered this blender last week and honestly... its NOT what i expected!! The motor is SO loud 😠 and it leaked all over my counter on day 2. customer service was helpful tho, they're sending a replacement. wouldn't recommend unless you're desperate lol 2/5 stars."
}

for name, text in samples.items():
    print(f"--- {name} ---")
    print(text)
    print()
```

```python
tokenised = {}

for name, text in samples.items():
    sentences = sent_tokenize(text)
    words = word_tokenize(text)
    tokenised[name] = {"sentences": sentences, "words": words}

    print(f"=== {name} ===")
    print(f"Sentence tokens ({len(sentences)}):")
    for i, s in enumerate(sentences, 1):
        print(f"  {i}. {s}")
    print(f"\nWord tokens ({len(words)}):")
    print(f"  {words}")
    print()
```

```python
no_stopwords = {}

for name, data in tokenised.items():
    words = data["words"]
    filtered = [w for w in words if w.lower() not in stop_words]
    no_stopwords[name] = filtered

    vocab_before = set(w.lower() for w in words)
    vocab_after = set(w.lower() for w in filtered)

    print(f"=== {name} ===")
    print(f"Before removal: {words}")
    print(f"After removal:  {filtered}")
    print(f"Vocabulary size before: {len(vocab_before)}")
    print(f"Vocabulary size after:  {len(vocab_after)}")
    print()
```

```python
# Punctuation/emoji symbols must be stripped before stemming/lemmatisation and
# vocabulary counting, because "amazing" and "amazing!!!" would otherwise be
# treated as different tokens, inflating vocabulary size purely from noise.
PUNCT_PATTERN = re.compile(r"[^\w\s]")

no_punct = {}

for name, filtered_words in no_stopwords.items():
    cleaned = [PUNCT_PATTERN.sub("", w) for w in filtered_words]
    cleaned = [w for w in cleaned if w.strip() != ""]
    no_punct[name] = cleaned

    print(f"=== {name} ===")
    print(f"Before: {no_stopwords[name]}")
    print(f"After:  {cleaned}")
    print()
```

```python
for name, words in no_punct.items():
    print(f"=== {name} ===")
    print(f"{'Original':<15}{'Stemmed':<15}{'Lemmatised':<15}")
    print("-" * 45)
    for w in words:
        print(f"{w:<15}{stemmer.stem(w):<15}{lemmatizer.lemmatize(w.lower()):<15}")
    print()
```

```python
for name in samples:
    print(f"##### {name} #####")
    print(f"0. Raw text:            {samples[name]}")
    print(f"1. Word tokens:         {tokenised[name]['words']}")
    print(f"2. Stopwords removed:   {no_stopwords[name]}")
    print(f"3. Punctuation removed: {no_punct[name]}")
    stemmed_list = [stemmer.stem(w) for w in no_punct[name]]
    lemma_list = [lemmatizer.lemmatize(w.lower()) for w in no_punct[name]]
    print(f"4a. Stemmed:            {stemmed_list}")
    print(f"4b. Lemmatised:         {lemma_list}")
    print("\n" + "="*80 + "\n")
```

## Why each preprocessing step matters

**Tokenisation** breaks raw text into words and sentences, the basic units every downstream NLP step operates on. Skipping it makes it impossible to reliably count words or align text with models expecting token sequences.

**Stopword removal** strips high-frequency, low-information words ("the", "is", "and") that dominate raw counts but carry little meaning. Skipping it lets these words drown out meaningful content in frequency-based models.

**Punctuation stripping** normalises tokens so "amazing" and "amazing!!!" are treated as the same word. Skipping it fragments the vocabulary and inflates token counts with noise.

**Stemming** crudely chops words to a root form (fast, sometimes produces non-words like "batteri"). Skipping it means inflected forms ("run", "running", "ran") are treated as unrelated, hurting recall in search/topic modelling.

**Lemmatisation** maps words to valid dictionary base forms using vocabulary knowledge — slower but linguistically accurate. Skipping it (or substituting stemming where lemmatisation is needed) can hurt precision in meaning-sensitive tasks like QA.

Together, these steps normalise noisy raw text (emojis, casing, punctuation, inflection) so downstream models see a clean, information-dense vocabulary.
