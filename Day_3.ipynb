```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string

raw = """Natural language processing transforms human text into structured
data that machines can learn from. Every pipeline begins here."""

# Step 1: Tokenize
tokens = word_tokenize(raw)
print("BEFORE:", raw)
print("\nAFTER tokenization:", tokens)
```

```python
stop_words = set(stopwords.words('english'))
punctuation = set(string.punctuation)

cleaned = [
    token.lower()
    for token in tokens
    if token.lower() not in stop_words
    and token not in punctuation
]

print("AFTER stop word + punctuation removal:", cleaned)
```

```python
from collections import Counter

bow = Counter(cleaned)
print("\nBag of Words:")
for word, count in bow.most_common():
    print(f"  '{word}': {count}")
```

```python
# Wrap it clean for reuse
def preprocess(text):
    tokens = word_tokenize(text.lower())
    stop_words = set(stopwords.words('english'))
    cleaned = [t for t in tokens if t not in stop_words and t not in string.punctuation]
    bow = Counter(cleaned)
    return cleaned, bow

paragraph = """
Machine learning models require clean, structured input. Raw text is messy —
it contains punctuation, filler words, and inconsistent casing.
Preprocessing removes the noise so the model sees only signal.
"""

tokens_clean, bow = preprocess(paragraph)

print("Cleaned tokens:", tokens_clean)
print("\nBag of Words:", dict(bow))
```

```python
import spacy
nlp = spacy.load("en_core_web_sm")

doc = nlp("Natural language processing transforms human text into structured data.")

tokens_clean = [
    token.lemma_.lower()
    for token in doc
    if not token.is_stop and not token.is_punct
]

print(tokens_clean)
# ['natural', 'language', 'processing', 'transform', 'human', 'text', 'structured', 'datum']
```
